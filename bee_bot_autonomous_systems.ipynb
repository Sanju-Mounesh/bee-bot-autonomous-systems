# Initial State and Goal State 

```python
import random as rnd 
boundaries = ["top", "bottom", "left", "right"] # Setting the boundary in the grid.
directions = ["N", "E", "S", "W"] # Giving directions 
initial_boundary = rnd.choice(boundaries) # To choose the states from this boundary 

initial_row = 0 
initial_column = 0 

if initial_boundary == "top":
    initial_column = rnd.randint(0, columns -1) # It will -1 and choose the the first column 
    initial_row = 0 # The topmost row in the grid 
    
elif initial_boundary == "bottom":
    initial_column = rnd.randint(0, columns -1) # It will -1 and choose the the first column
    initial_row = rows-1 # -1 max row entered gives the last row of the grid.

elif initial_boundary == "left": 
    initial_row = rnd.randint(0, rows -1)
    initial_column = 0 #The first column in the grid. 
    
elif initial_boundary == "right":
    initial_row = rnd.randint(0, rows -1)
    initial_column = columns -1 # -1 max columns entered gives the last column of the grid. 

if initial_boundary == "top": # Defining the directions referring the boundary
    initial_direction = "S"  # When the initial state is in top it will enter in the direction of south
elif initial_boundary == "bottom": # Defining the directions referring the boundary 
     initial_direction = "N" # When the initial state is in bottom it will enter in the direction of North.

elif initial_boundary == "left": # Defining the directions referring the boundary 
     initial_direction = "E" # When the initial state is in left it will enter in the direction of East.


elif initial_boundary == "right": # Defining the directions referring the boundary 
     initial_direction = "W" # When the initial state is in right it will enter in the direction of West.

initial_state = (initial_row, initial_column, initial_direction)# Combines the row, column and direction
```

According to the game. In the above cell the Initial State will be randomly generated inside the boundary.
Depending of the selected boundary the if and elif conditions will choose a random state.
The initial state represents initial_row, initial_column, initial_direction.
Also in the above cell it is defined how the Bee-Bot should enter the grid according to the initial State.

#Goal State 

```python
import random as rnd

#This is to reassign the goal state if initial state and goal state are common 
goal_row = initial_row
goal_column = initial_column

while (goal_row, goal_column) == (initial_row, initial_column): #To ensure Initial State and Goal State should not be same 
    goal_boundary = rnd.choice(boundaries) # To choose the state from this boundary
    goal_row = 0 #
    goal_column = 0
    
    if goal_boundary == "top": # Assigning the row 0 as the boundary 
        goal_column = rnd.randint(0, columns -1) # 
        goal_row = 0 

    elif goal_boundary == "bottom": # Assigning the max entered row -1  as the boundary
        goal_column = rnd.randint(0, columns -1)
        goal_row = rows-1

    elif goal_boundary == "left": # Assigning the column 0 as the boundary
        goal_row = rnd.randint(0, rows -1)
        goal_column = 0

    elif goal_boundary == "right": # Assigning the max entered column -1  as the boundary
        goal_row = rnd.randint(0, rows -1)
        goal_column = columns-1
    
    if goal_boundary == "top": # Assigning the direction for the goal as required 
        goal_direction = "S"

    elif goal_boundary == "bottom": # Assigning the direction for the goal as required
        goal_direction = "N"

    elif goal_boundary == "left": # Assigning the direction for the goal as required
        goal_direction = "E"

    elif goal_boundary == "right": # Assigning the direction for the goal as required
        goal_direction = "W"
        
        
    goal_state = (goal_row, goal_column, goal_direction)
```

In the above cell we have defined that Initial state and Goal state should not be the same.

# Random Obstacles 

```python
import random as rnd 
total_cells = (rows * columns) # Multiplies the rows and columns entered by the user.
random_obstacles = int(total_cells * 0.20) # Adds 20% of obstacles to the Grid.
#print(total_cell)
#print(random_obstacles)

obstacle = "X" # "X" represents the obstacles in the grid 

placed_obstacles = 0 # Counts how many obstacles have been successful.

while placed_obstacles < random_obstacles:
    obstacle_row = rnd.randint(0, rows -1) #Selects random rows 
    obstacle_column = rnd.randint(0, columns -1) # Selects random columns
    
    if ((obstacle_row, obstacle_column) != initial_state[:2] # Avoids an obstacle from blocking the initial state 
        and (obstacle_row, obstacle_column) != goal_state[:2] # Avoids an obstacle from blocking the goal state
        and gameState[obstacle_row][obstacle_column] == " "): # Ensures the selected cell is currently empty 
        
        gameState[obstacle_row][obstacle_column] = obstacle
        placed_obstacles += 1 # The selected empty cell is replaced with "X"

print(gameState)
```

The above Random Obstacles cell helps to generate the obstacles and to select the total percentage of obstacles required, Random positions of the obstacles,
Also prevents blocking of Bee-Bot to go through initial state and goal state by the obstacles.
Upgrades the obstacles in empty cell as "X".

# Actions

```python
#Action and Validation 
#Actions 
# Actions for the Bee-Bot 
actions = [ "UP", "DOWN", "LEFT", "RIGHT" ]

def apply_action(state, action):
    row, column, direction = state
    new_row = row
    new_column = column
    current_index = directions.index(direction)#current_directions = directions[0,1,2,3]
    new_direction = direction # finding out the new index value of direction
    
    if action == "RIGHT":
        new_index = (current_index + 1 )% 4 # Action to turn right 
        # (current index value of direction + 1 value) % 4 available actions 
        new_direction = directions[new_index]
    elif action == "LEFT":
        new_index = (current_index - 1 )% 4 # Action to turn left 
         # (current index value of direction + 1 value) % 4 available actions 
        new_direction = directions[new_index]

        
    elif action == "UP": # action to move forward 
        if direction == "N": # if the Bee-Bot is facing North the upward movement will decrease the rows
            new_row = state[0] - 1 # Here state[0] represents the rows 
        elif direction == "E": # if the Bee-Bot is facing East the upward movement will increase the columns
            new_column = state[1] + 1 # Here state[1] represents the columns
        elif direction == "S":# if the Bee-Bot is facing North the upward movement will increase the rows
            new_row = state[0] + 1 # Here state[0] represents the rows
        elif direction == "W":  # if the Bee-Bot is facing East the upward movement will decreases the columns
            new_column = state[1] - 1 # Here state[1] represents the columns

        
            
    elif action == "DOWN":
         if direction == "N": # if the Bee-Bot is facing North the upward movement will increase the rows
            new_row = state[0] + 1 # Here state[0] represents the rows 
         elif direction == "E": # if the Bee-Bot is facing East the upward movement will decrease the columns
            new_column = state[1] - 1 # Here state[1] represents the columns
         elif direction == "S":# if the Bee-Bot is facing North the upward movemnet will decrease the rows
            new_row = state[0] - 1 # Here state[0] represents the rows
         elif direction == "W":  # if the Bee-Bot is facing East the upward movemnet will increase the columns
            new_column = state[1] + 1 # Here state[1] represents the columns
       
 
    return (new_row, new_column, new_direction)   
```

The above cell defined the actions for the Bee-Bot to reach the goal. The Bee-Bot has four actions Up, Down, Left, Right.
Up actions moves the Bee-Bot forward, Down actions moves the Bee-Bot backward, Left will turn the Bee-bot left and Right will turn the Bee-Bot right.
apply_action uses the current position and direction to determine the next state. After applying the action the function will update the new state of the Bee-Bot.

# Action Validation

Bee-Bot should not move beyond row 0 

Bee-Bot should not go through the obstacles 

Bee-Bot should only move through the empty cell 

```python
def is_valid_action(state, action):
    new_state = apply_action(state, action) # Calculates the state that would result from selected action
    new_row, new_column, new_direction = new_state

    if new_row < 0 or new_row >= rows: #Checks whether the new row is outside the grid
        return False
    
    if new_column < 0 or new_column >= columns: # Checks whether the new columns is outside the grid
        return False

    if gameState[new_row][new_column] == "X": # Checks the obstacles if outside the grid.
        return False 

    return True 
```

# DFS (Depth-First Search)

```python
stack = []
stack.append(initial_state) 
#Creates an empty stack and adds the initial state

visited = set()
visited.add(initial_state) # Records the initial state as already visited 
#Creates a set to remember states we have already explored

parent = {} #Stores previous states 
parent[initial_state] = None

found = False

import time 
start_time = time.time()

while stack != []: 
    current_state = stack.pop() # Removes the most recently added state
    
    if current_state == goal_state:
        print("Goal found!")
        found = True
        break
        
    for action in actions:
        if is_valid_action(current_state, action): #Checks whether the each action is allowed from current state.
            next_state = apply_action(current_state, action)
            
            if next_state not in visited:
                visited.add(next_state)
                parent[next_state] = current_state
                stack.append(next_state)

end_time = time.time()
execution_time = end_time - start_time

print("DFS execution time:", execution_time, "seconds")                

print("Goal found status:", found)
print("Number of visited states:", len(visited))
```

DFS searches a path from initial state to goal state. Stack uses the LIFO exploring the recent added safest first. remembers the visited and parent path and prevents repeating.
The search stops when the goal is reached. It shows the execution time and number of recorded states are displayed.

```python
# To verify the DFS
print("Initial state:", initial_state)
print("Goal state:", goal_state)

for action in actions:
    print("Action:", action)
    print("New state:", apply_action(initial_state, action))
    print("Valid:", is_valid_action(initial_state, action))
```

```python
if found:
    path = []
    current = goal_state
    
    while current is not None:
        path.append(current) # adding current state to the path
        current = parent[current]
    path.reverse() # Cause the recorded path need to be reversed accordingly
    print("DFS Solution Path:")
    print(path)
    print("Number of states:", len(path))

else:
    print("No path found!")
```

The above cell adds the current state to ppath, Looks at the previous state using the parent dictionary
repeats until it reaches none. Our parent dictionary stores the previous state for each state discovered during DFS. The path later reveres the list in the correct order initial state to goal state.

# Heuristic Function Design 

```python
# Manhattan distance 
def heuristic(state, goal_state):
    current_row = state[0]
    current_column = state[1]
    goal_row = goal_state[0]
    goal_column = goal_state[1]

    distance = abs(current_row - goal_row) + abs(current_column - goal_column)
    return distance 
# Manhattan distance heuristic estimates the minimum actual cost to reach
```

In the above ceell Heuristic Function Design we have used Manhattan Distance becuase this does not overestimate the minimum number of actions needed to reach the goal.
It ignores obstacles and turning cosrs, so it can undereestimate the actual cost. This helps A* prioritize states that are closer to the goal.

# A* Search Implementation

```python
open_list = [] 

g_cost = 0 # zero cause the Bee-Bot has not moved 
h_cost = heuristic(initial_state, goal_state) # Estimated distance to reaach the goal 
f_cost = (g_cost + h_cost) # total cost 

open_list.append([initial_state, g_cost, h_cost, f_cost]) # This will place the starting state to begin searching

visited = set() #Keeps record of states already explored 
parent = {} # remembers where each state came from 
parent[initial_state] = None # The initial state has no parent 

# Adding this dictionary to know which route is cheaper  
g_costs = {}
g_costs[initial_state] = 0 #Cheapest cost found is added in this dictionary 
found = False # Initial state will find goal so its False 

import time 
start_time = time.time()

while open_list != []: # continues searcch while there are states required to explore 
        min_index = 0
    
        for i in range(1, len(open_list)):
            if open_list[i][3] < open_list[min_index][3]:
                min_index = i
            
        current_entry = open_list.pop(min_index)#pop(min_index) removed the entry with lowest f-cost

        current_state, current_g, current_h, current_f = current_entry # Bee-Bot current position and direction
        
        if current_g > g_costs[current_state]:
            continue # First checks the selected entry if outdated and skips and moves to the next iteration 

        if current_state == goal_state:
            print("Goal found!")
            found = True #A* finds the goal
            break
                
        for action in actions:
            if is_valid_action(current_state, action): # Avoids the Bee-Bot moving outside the grid or obstacles
                next_state = apply_action(current_state, action) # Calculates the reuslts of position and direction
                tentative_g = current_g + 1 # To calculate the possible cost of reaching the next state
                
                if next_state not in g_costs or tentative_g < g_costs[next_state]: # Checks weather never reached this state before
                    g_costs[next_state] = tentative_g # to remember the new cheapest cost 
                    parent[next_state] = current_state # to remember which state we came from
                    h_cost = heuristic(next_state, goal_state) # total travelled, goal estimation, estimated total cost 
                    f_cost = tentative_g + h_cost
                    open_list.append([next_state, tentative_g, h_cost, f_cost])# Adds the new state and costs to the remaning list waiting to be explored

print("Number of States Recorded:", len(g_costs)) # to know how many states have been recorded 
end_time = time.time()
execution_time = end_time - start_time

print("A* execution time:", execution_time, "seconds")
```

A* uses open_list to store states waiting to be explored. The initial state's g-cost is zero and the heuristic function calculates its h-cost. The fcost is calculated by adding g-cost and h-cost.The initial state is added to the open list. Here while open_list continues searching while thee are states waiting to be explored.min_index=0 initially assumes first entry has the lowest f-cost. 'for' loop comapres the f-cost of the entries in the open list and later open_list[i][3] access the f-cost which is the fourth element in each entry. min_index = i remembers the position of the entry with lowest f-cost and later removes and selects that entry for exploration using pop(min_index).

A* compares the selected entries g-xost with the best known cost stored in g-costs if the selected entry has a higher cost, continue skips it because a cheaper route has already been found.

The algorithm compares the current state with the goal state if they match it sets found = true and uses break to stop the search because the destination has been reached. 

A* explored all four avilable actions and ensures an actio does not move the Bee-Bot outside the grid or into a obstacle. If the next state is new or a cheaper route is found A* updates its recorded cost,parent,heuristic and f-cosr then adds updated entry to the open list.

```python
if found:
    path = [] # to create an empty list of the soultion
    current = goal_state # to start from the destination

    while current is not None: # To stop the loop when reaches the initial state
        path.append(current) # to store the  current state
        current = parent[current] # to move one step backwards 

    path.reverse() # To reverse the backward collected values

    print("A* Solution Path:")
    print(path)
    print("Number of states:", len(path))

else:
    print("No path found!")
```

The above cell checks whether A* successfully found the goal, Creates an empty list called path to store the solution states. It starts from goal_state and follows the parent dictionary to move backwards through the states.Each state is added tot the path using path.aapend(current). 
The loop stops when it reaches None. Fianlly it prints the solution path and the total number of states.
